## Python + Google Sheets com Apps Script

Este notebook envia e consulta mensagens (nome e e-mail) em uma **planilha do Google**, usando o **Google Apps Script** como uma pequena API web.

## Como usar este notebook

Execute as células **de cima para baixo**: clique no botão de executar à esquerda de cada célula ou use `Shift + Enter`. O Colab já traz o Python e a biblioteca `requests`, então não é preciso instalar nada.

## 1. Configuração

Importamos as bibliotecas e definimos a `API_URL`, o endereço do seu Apps Script.

- `requests`: faz as chamadas HTTP (POST e GET).
- `json`: formata a resposta para ficar legível na tela.

**Troque o valor de `API_URL` pela URL da sua implantação**, mantendo as aspas.

```python
import json

import requests

# Cole aqui a URL do seu aplicativo da web (termina em /exec).
API_URL = "https://script.google.com/macros/s/AKfycbxGWTbi9K4wLZFOpSAjk13hR-mY86Gu94jh9VyelStMf1004mlxYxmvW65BrXQpV-e1Kw/exec"
# Aviso útil: se a URL ainda não foi trocada, o código abaixo não vai funcionar.
if not API_URL.startswith("https://script.google.com/"):
    print("Atenção: troque a API_URL pela URL da sua implantação do Apps Script.")
else:
    print("API_URL configurada.")
```

## 2. Funções de envio

Estas funções mandam dados **para** a planilha, usando uma requisição **POST**. O Apps Script recebe o pedido na função `doPost` e grava uma nova linha.

| Função | O que faz |
|---|---|
| `enviar_mensagem(nome, email)` | Envia **uma** mensagem e imprime a resposta do servidor |
| `enviar_varias_mensagens(mensagens)` | Recebe uma **lista** de mensagens e envia uma por uma |

```python
def enviar_mensagem(nome, email):
    """Envia uma mensagem (nome e e-mail) para a planilha.

    Faz um POST para a API_URL com os dados em formato JSON.
    O Apps Script grava uma nova linha e responde com um texto de confirmação,
    que é impresso na tela.
    """
    resposta = requests.post(
        API_URL,
        json={"nome": nome, "email": email},  # as chaves precisam ser "nome" e "email"
    )
    print(resposta.text)


def enviar_varias_mensagens(mensagens):
    """Envia várias mensagens, uma de cada vez.

    Parâmetro:
        mensagens: lista de dicionários no formato
                   {"nome": "...", "email": "..."}
    """
    for mensagem in mensagens: # complete esta linha
        enviar_mensagem(mensagem['nome'], mensagem['email'])
```

## 3. Funções de consulta

Estas funções buscam dados **da** planilha, usando uma requisição **GET**. O Apps Script responde na função `doGet`, devolvendo todas as linhas em formato **JSON**: uma lista de objetos, em que cada objeto é uma linha e as chaves vêm dos cabeçalhos da planilha.

| Função | O que faz |
|---|---|
| `obter_todas_mensagens()` | Busca e imprime **todas** as mensagens |
| `obter_ultima_mensagem()` | Busca e imprime **somente a última** (ou avisa se não houver nenhuma) |

```python
def obter_todas_mensagens():
    """Busca todas as mensagens da planilha e imprime em JSON formatado."""
    resposta = requests.get(API_URL) # complete esta linha
    dados = resposta.json() # complete esta linha
    print(json.dumps(dados, indent=2, ensure_ascii=False))


def obter_ultima_mensagem():
    """Busca as mensagens da planilha e imprime apenas a última."""
    resposta = requests.get(API_URL) # complete esta linha
    dados = resposta.json()
    if dados:
        ultima = dados[-1] # complete esta linha
        print(json.dumps(ultima, indent=2, ensure_ascii=False))
    else:
        print("Nenhuma mensagem encontrada.")
```

## 4. Executando

Agora vamos usar as funções, uma etapa por célula. Depois de cada etapa, vale **conferir a planilha**: as linhas novas aparecem abaixo dos cabeçalhos.

### 4.1 Enviar uma mensagem

Deve imprimir a confirmação do servidor (`Sucesso`).

```python
print("Enviando uma mensagem...")
enviar_mensagem("alice", "alice@example.com")
```

### 4.2 Enviar várias mensagens

Uma lista de dicionários, enviada de uma vez. Deve imprimir uma confirmação por mensagem.

```python
print("Enviando várias mensagens...")
mensagens = [
    {"nome": "bob", "email": "bob@example.com"},
    {"nome": "charlie", "email": "charlie@example.com"},
]
enviar_varias_mensagens(mensagens)
```

### 4.3 Buscar todas as mensagens

Deve imprimir a lista completa em JSON. As chaves seguem os cabeçalhos da planilha (`Data`, `Nome` e `Email`).

```python
print("Buscando todas as mensagens...")
obter_todas_mensagens()
```

### 4.4 Buscar a última mensagem

Deve imprimir somente a linha mais recente.

```python
print("Buscando a última mensagem...")
obter_ultima_mensagem()
```
